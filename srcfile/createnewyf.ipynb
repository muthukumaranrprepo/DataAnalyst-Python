```python
# Debt to Equity ratio
import pandas as pd
import yfinance as yf

# Format display globally to convert raw numbers and suppress scientific notation
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

ticker = yf.Ticker("SHARDACROP.NS")
balance_sheet = ticker.get_balance_sheet()

# 1. Extract rows using exact yfinance index names (with spaces)
total_assets = balance_sheet.loc["TotalAssets"]
total_liabilities = balance_sheet.loc["TotalLiabilitiesNetMinorityInterest"]
common_stock = balance_sheet.loc["CommonStock"]

# 2. Accounting Calculations
total_equity = total_assets - total_liabilities  # Net Worth
shareholderequity = total_equity + common_stock           # Reserves & Surplus

DebttoEquityRatio = total_liabilities / shareholderequity

# 3. Combine into a clean DataFrame for side-by-side historical display
summary_df = pd.DataFrame(
    {
        "Total Assets": total_assets,
        "Total Liabilities": total_liabilities,
        "Reserves & Surplus": total_equity,
        "Common Stock": common_stock,
        "Shareholder Equity": shareholderequity,
        "Debt-to-Equity Ratio": DebttoEquityRatio,
    }
).T

print("=== SHARDA CROPCHEM BALANCE SHEET BREAKDOWN (in INR) ===")
print(summary_df)

```

```python
import pandas as pd
import yfinance as yf

# Format numbers with commas and 2 decimal places
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")

ticker = yf.Ticker("SHARDACROP.NS")
balance_sheet = ticker.get_balance_sheet()

# 1. Extract rows directly as Pandas Series using .loc
total_debt = balance_sheet.loc["TotalDebt"]
total_equity = balance_sheet.loc["TotalEquityGrossMinorityInterest"]

# 2. Select position 0 (most recent year) using .iloc
latest_debt = total_debt.iloc[0]
latest_equity = total_equity.iloc[0]

# 3. Calculate Debt-to-Equity Ratio
de_ratio = latest_debt / latest_equity

# Print results
latest_date = balance_sheet.columns[0].strftime("%Y-%m-%d")
print(f"--- INFY Debt-to-Equity Ratio ({latest_date}) ---")
print(f"Total Debt:                        {latest_debt:,.2f}")
print(f"Total Equity:                      {latest_equity:,.2f}")
print(f"Debt-to-Equity Ratio:              {de_ratio:.4f}")
if de_ratio < 0.8 :
    print("Debt-to-Equity Ratio is below 0.8, Durable advantage company with low debt risk.")
else:
    print("Debt-to-Equity Ratio is above 0.8, High debt risk company with low durable advantage.")
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup


def get_screener_financials(ticker_symbol):
    # Screener URL
    url = f"https://www.screener.in/company/{ticker_symbol}/consolidated/"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
    }

    response = requests.get(url, headers=headers)

    # Fallback to standalone financials if consolidated is unavailable
    if response.status_code != 200 or "Page not found" in response.text:
        url = f"https://www.screener.in/company/{ticker_symbol}/"
        response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.content, "html.parser")

    def parse_screener_table(section_id):
        section = soup.find("section", {"id": section_id})
        if not section:
            return None

        table = section.find("table")
        if not table:
            return None

        # 1. Extract date headers (e.g., ['Mar 2016', 'Mar 2017', ... 'TTM'])
        th_tags = table.find_all("th")
        date_headers = [th.text.strip() for th in th_tags if th.text.strip()]

        # 2. Extract row titles and values
        rows = []
        for tr in table.find_all("tr"):
            tds = tr.find_all("td")
            if not tds:
                continue

            # First <td> is the metric name (e.g., Sales, Expenses)
            # Remove '+' symbols from expandable rows
            row_title = tds[0].text.replace("+", "").strip()

            # Remaining <td>s are numerical values
            row_values = [
                td.text.strip().replace(",", "") for td in tds[1:]
            ]

            if row_title and row_values:
                rows.append([row_title] + row_values)

        if not rows:
            return None

        # 3. Align columns: 'Metric' + 12 Date Headers
        col_names = ["Metric"] + date_headers

        # Create and index DataFrame
        df = pd.DataFrame(rows, columns=col_names)
        df.set_index("Metric", inplace=True)

        return df

    pnl = parse_screener_table("profit-loss")
    bs = parse_screener_table("balance-sheet")
    cf = parse_screener_table("cash-flow")

    return pnl, bs, cf


# Test with RELIANCE, INFY, or TCS
pnl, bs, cf = get_screener_financials("RELIANCE")

print("--- 10-Year Income Statement (Profit & Loss) ---")
print(pnl)
print(bs)
print(cf)
```
