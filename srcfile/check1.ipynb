```python
#import appointment inperson file:

import pandas as pd
df1 = pd.read_excel(r"C:\Users\muthu\Downloads\Appointment_Inperson.xlsx",
                   usecols=['Date','Name','Patient ID','Treatment Type','Consultant','Follow Up',
                            'Created By','Call Center Agent','Receptionist'])
df2 = pd.read_excel(r"C:\Users\muthu\Downloads\Appointment_Online.xlsx",
                   usecols=['Date','Name','Patient ID','Treatment Type','Consultant','Follow Up',
                            'Created By','Call Center Agent','Receptionist'])
```

```python
print('Before Inperson:',df1.size)
# Drops rows where 'column_name' is spaces, empty strings, or NaN
df1 = df1[df1['Patient ID'].str.strip().fillna('') != '']
print('After INperson:',df1.size)

print('Before Online:',df2.size)
# Drops rows where 'column_name' is spaces, empty strings, or NaN
df2 = df2[df2['Patient ID'].str.strip().fillna('') != '']
print('After Online:',df2.size)
```

```python
with pd.ExcelWriter(r"C:\Users\muthu\Downloads\Appointment_Cons.xlsx", engine='openpyxl') as writer:
    df1.to_excel(writer, sheet_name='Inperson', index=False)
    df2.to_excel(writer, sheet_name='Online', index=False)
```

```python


# Concatenate vertically
df_combined = pd.concat([df1, df2], ignore_index=True)
# Write to Excel
df_combined.to_excel(r"C:\Users\muthu\Downloads\Appointment_Consolidated.xlsx", sheet_name="Combined", index=False)
```
