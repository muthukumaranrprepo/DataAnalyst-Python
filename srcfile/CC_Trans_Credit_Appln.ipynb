```python
# credit card transaction application record data analysis
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#appln_df=pd.read_csv(r'https://github.com/muthukumaranrprepo/DataAnalyst-Python/blob/main/DataFiles/CC_Tran_application_record.csv?raw=true')
appln_df=pd.read_csv('C:\\Users\\hp\\Documents\\GitHub\\DataAnalyst-Python\\DataFiles\\CC_Tran_application_record.csv')
appln_df.info()
```

```python
#credit_df = pd.read_csv(r'https://github.com/muthukumaranrprepo/DataAnalyst-Python/blob/main/DataFiles/CC_Tran_credit_record.csv?raw=true')
credit_df = pd.read_csv(r'C:\Users\hp\Documents\GitHub\DataAnalyst-Python\DataFiles\CC_Tran_credit_record.csv')
credit_df.info()
```

```python
#Plot for Null values - Identify the missing values through visualization
#appln_df.isnull().sum()
missingdata = pd.DataFrame(appln_df.isnull().sum()/len(appln_df)*100).reset_index()
plt.figure(figsize=(30,5))
sns.pointplot(data=missingdata, x='index',y=0)
```

```python
#identify the list of numerical columns in the dataframe
appln_df.describe().T
```

```python
# Plot for Null values - Identify the missing values through visualization
missing_data_credit = pd.DataFrame(credit_df.isnull().sum() / len(credit_df) * 100).reset_index()
plt.figure(figsize=(30, 5))
sns.pointplot(data=missing_data_credit, x='index', y=0)
plt.title('Percentage of Missing Values in Credit DataFrame')
plt.xlabel('Columns')
```
